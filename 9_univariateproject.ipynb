# Renaming Columns.

```python
df=pd.DataFrame(file)

df.rename(columns={'Product line': 'product_line'}, inplace=True)
df.rename(columns={'Invoice ID': 'invoiceID'}, inplace=True)
df.rename(columns={'Customer type': 'customertype'}, inplace=True)
df.rename(columns={'Unit price': 'unit_price'}, inplace=True)
df.rename(columns={'Tax 5%': 'tax_5'}, inplace=True)
df.rename(columns={'gross margin': 'gross_margin'}, inplace=True)
df.rename(columns={'gross income': 'gross_income'}, inplace=True)
df.head()
```

This cell first creates a new DataFrame `df` as a copy of `file`. Then, it renames several columns in the `df` DataFrame for easier access and consistency by converting them to snake_case. Finally, it displays the first 5 rows of the modified DataFrame using `df.head()`.

### Analyzing Sales Distribution by Payment Method

```python
#file.Payment.value_counts()
payment_mode_count = file.groupby('Payment')['Total'].sum()
X = payment_mode_count.index
Y = payment_mode_count.values
pl.pie(Y ,labels= X,autopct='%1.2f%%')
pl.axis("equal")
pl.show()
```
